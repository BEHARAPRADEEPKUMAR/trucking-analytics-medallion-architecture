```python
# Import required functions for Bronze loading
from delta.tables import DeltaTable
from pyspark.sql.functions import col, current_timestamp


# Reusable function to load a DataFrame into the Bronze Delta layer
def load_to_bronze(df, schema, bronze_path, merge_condition):

    # Apply the predefined schema to the DataFrame
    df_bronze = df.select([
        col(field.name).cast(field.dataType).alias(field.name)
        for field in schema.fields
    ])

    # Add ingestion timestamp
    df_bronze = df_bronze.withColumn(
        "_ingested_at",
        current_timestamp()
    )

    # Create the Bronze Delta table if it does not exist
    if not DeltaTable.isDeltaTable(spark, bronze_path):

        df_bronze.write \
            .format("delta") \
            .save(bronze_path)

    # If Bronze table already exists, update it using MERGE
    else:

        bronze_table = DeltaTable.forPath(
            spark,
            bronze_path
        )

        (
            bronze_table.alias("target")
            .merge(
                df_bronze.alias("source"),
                merge_condition
            )
            .whenMatchedUpdateAll()
            .whenNotMatchedInsertAll()
            .execute()
        )

    # Display a sample of the Bronze table
    display(
        spark.read
        .format("delta")
        .load(bronze_path)
        .limit(30)
    )
```

```python
# Define the Bronze storage location
bronze_base = "/Volumes/workspace/default/trucking_analytics/bronze"
# Load Customers into Bronze
load_to_bronze(
    customers_df,
    customers_schema,
    f"{bronze_base}/customers",
    "target.customer_id = source.customer_id"
)
```
